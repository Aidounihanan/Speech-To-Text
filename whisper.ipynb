```python
import whisper

model = whisper.load_model("small")

# Transcrire avec détection automatique de la langue
darija_text = model.transcribe("audio3.opus", language="ar")["text"]
fr_text = model.transcribe("audio3.opus", language="fr")["text"]

print("🟢 Version en Darija/Arabe :", darija_text)
print("🔵 Version en Français :", fr_text)
```

```python
import whisper

# Charger les modèles
model_small = whisper.load_model("small")
model_medium = whisper.load_model("medium")
model_large = whisper.load_model("large")

# Transcrire avec chaque modèle
result_small = model_small.transcribe(
    "audio3.opus",
    language="fr",
    temperature=0.0,
    beam_size=5,
    word_timestamps=True,
    condition_on_previous_text=True
)

result_medium = model_medium.transcribe(
    "audio3.opus",
    language="fr",
    temperature=0.0,
    beam_size=5,
    word_timestamps=True,
    condition_on_previous_text=True
)

result_large = model_large.transcribe(
    "audio3.opus",
    language="fr",
    temperature=0.0,
    beam_size=5,
    word_timestamps=True,
    condition_on_previous_text=True
)

# Afficher les résultats de chaque modèle
print("Résultat du modèle small:")
print(result_small["text"])

print("\nRésultat du modèle medium:")
print(result_medium["text"])

print("\nRésultat du modèle large:")
print(result_large["text"])
```
