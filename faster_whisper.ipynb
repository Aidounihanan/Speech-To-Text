```python
from faster_whisper import WhisperModel

model = WhisperModel("large-v3")
```

```python
# Transcrire un fichier audio avec détection automatique de la langue
def transcribe_audio(audio_path):
    # Transcription avec détection automatique de la langue
    segments, _ = model.transcribe(audio_path, beam_size=5)

    # Retourner le texte de la transcription
    transcript = " ".join([segment.text for segment in segments])
    return transcript

# Exemple de transcription d'un fichier audio
audio_file = "/content/monAudio.wav.mp3"
transcription = transcribe_audio(audio_file)
print("Transcription:", transcription)
```
