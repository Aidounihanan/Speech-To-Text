<h2>Wav2Vec</h2>


```python
import torch
import torchaudio
import librosa
import soundfile as sf
from transformers import Wav2Vec2Processor, Wav2Vec2ForCTC

# Charger le modèle pré-entraîné Wav2Vec 2.0 pour le français
model_name = "facebook/wav2vec2-large-xlsr-53-french"
processor = Wav2Vec2Processor.from_pretrained(model_name)
model = Wav2Vec2ForCTC.from_pretrained(model_name)

# Charger et prétraiter l'audio (convertir en 16kHz si nécessaire)
def load_audio(file_path):
    speech, sr = librosa.load(file_path, sr=16000)  # Resampling à 16kHz
    return speech

# Transcrire un fichier audio
def transcribe_audio(file_path):
    speech = load_audio(file_path)

    # Préparer l'entrée pour le modèle
    inputs = processor(speech, sampling_rate=16000, return_tensors="pt", padding=True)
    with torch.no_grad():
        logits = model(inputs.input_values).logits

    # Décoder la transcription
    predicted_ids = torch.argmax(logits, dim=-1)
    transcription = processor.batch_decode(predicted_ids)[0]

    return transcription

# Chemin du fichier audio (WAV format, 16kHz recommandé)
audio_file = "chemin/vers/votre_fichier.wav"

# Exécution de la transcription
transcription = transcribe_audio(audio_file)
print("Transcription :", transcription)
```
